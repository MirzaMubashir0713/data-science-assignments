```python
""" ESTIMATION AND CONFIDENCE INTERVAL """
import numpy as np
import scipy.stats as stats

# Data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

print("------------------------ Task A: 99% CI Using Sample Std Dev ------------------------")
# Rationale: Small sample size (n = 15 < 30) and unknown population std dev -> Use t-distribution
df = n - 1
t_crit = stats.t.ppf(0.995, df)
margin_of_error_t = t_crit * (s / np.sqrt(n))
ci_t = (mean - margin_of_error_t, mean + margin_of_error_t)

print(f"Sample Mean: {mean:.4f}")
print(f"Sample Standard Deviation (s): {s:.4f}")
print(f"Degrees of Freedom: {df}")
print(f"t-critical value (99%): {t_crit:.4f}")
print(f"99% Confidence Interval: ({ci_t[0]:.4f}, {ci_t[1]:.4f})")

print("\n------------------------ Task B: 99% CI Using Known Pop Std Dev ------------------------")
# Known population standard deviation (sigma = 0.2) -> Use z-distribution
sigma = 0.2
z_crit = stats.norm.ppf(0.995)
margin_of_error_z = z_crit * (sigma / np.sqrt(n))
ci_z = (mean - margin_of_error_z, mean + margin_of_error_z)

print(f"Population Standard Deviation (sigma): {sigma}")
print(f"z-critical value (99%): {z_crit:.4f}")
print(f"99% Confidence Interval: ({ci_z[0]:.4f}, {ci_z[1]:.4f})")
```

```python
""" HYPOTHESIS TESTING """ 
import numpy as np
import scipy.stats as stats

print("------------------------ Hypothesis Testing: Bombay Hospitality Ltd. ------------------------")

# Given parameters
mu_X = 600
sigma_X = 25
n = 25
alpha = 0.05

# 1. Theoretical Mean Cost (mu) based on W = 1000 + 5X
mu_W = 1000 + 5 * mu_X
sigma_W = 5 * sigma_X
standard_error = sigma_W / np.sqrt(n)

# Sample mean given in claim investigation (adjusted to reflect 'higher' operating costs: 4050 instead of typo 3050)
sample_mean = 4050 

print(f"Theoretical Mean Cost (mu): {mu_W}")
print(f"Standard Deviation of Cost (sigma_W): {sigma_W}")
print(f"Standard Error (SE): {standard_error}")

print("\n------------------------ Task 1: Hypotheses Statement ------------------------")
print("Null Hypothesis (H0): mu <= 4000 (Weekly operating cost is equal to or less than the model suggests)")
print("Alternative Hypothesis (H1): mu > 4000 (Weekly operating costs are higher than the model suggests)")

print("\n------------------------ Task 2: Calculate the Test Statistic ------------------------")
z_stat = (sample_mean - mu_W) / standard_error
print(f"Calculated Z-statistic: {z_stat:.4f}")

print("\n------------------------ Task 3: Determine Probability & Compare ------------------------")
z_crit = stats.norm.ppf(1 - alpha)
p_value = 1 - stats.norm.cdf(z_stat)
print(f"Critical Z-value (alpha = 0.05, one-tailed): {z_crit:.4f}")
print(f"P-value: {p_value:.4f}")

print("\n------------------------ Task 4: Make a Decision ------------------------")
if z_stat > z_crit:
    print("Decision: Reject the Null Hypothesis (H0) because Z-statistic > Critical Z-value (and p-value < alpha).")
else:
    print("Decision: Fail to reject the Null Hypothesis (H0).")

print("\n------------------------ Task 5: Conclusion ------------------------")
print("Conclusion: There is strong statistical evidence at the 5% significance level to support the restaurant owners' claim that the weekly operating costs are higher than the theoretical model suggests.")
```
