```python
#Exercise 1: Prime Numbers

num = int(input("Enter a number: "))
is_prime = True

if num <= 1:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print(num, "is a prime number.")
else:
    print(num, "is not a prime number.")
```

```python
# Exercise 2: Product of Random Numbers
import random

# Generate two random numbers
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Ask the user for the product
user_answer = int(input(f"What is {num1} x {num2}? "))

# Check if the answer is correct
correct_answer = num1 * num2

if user_answer == correct_answer:
    print("Correct! Good job!")
else:
    print(f"Incorrect. The correct answer was {correct_answer}.")
```

```python
#Exercise 4:
#Problem Statement (Question): Hospital Billing System

# Hospital Billing System
# Given patient details
days_stayed = 4
room_charge_per_day = 2000
doctor_fee = 1500
num_lab_tests = 3
lab_test_cost = 300
medicine_charges = 2400

# Calculating charges
room_charges = days_stayed * room_charge_per_day
lab_charges = num_lab_tests * lab_test_cost

# Total bill before discount
total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

# Apply 10% discount if total bill exceeds ₹10,000
if total_bill > 10000:
    discount = total_bill * 0.10
    final_bill = total_bill - discount
else:
    final_bill = total_bill

print("Total Bill (Before Discount): ₹", total_bill)
print("Final Bill (After Discount): ₹", final_bill)
```
