```python
# Section C: Exploring Pandas 
''' Section C: Exploring Pandas

11. Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then: 
- Use .info(), .describe() 
- Select 'Name' and 'City' columns 
- Drop 'City' column - Fill any missing values in 'Salary' column with the mean 
- Remove any duplicate rows
'''
import pandas as pd
import numpy as np

print("------------------------ Question - 11 ------------------------")

# Create a DataFrame with 10 rows, including a missing value in Salary and a duplicate row
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ian', 'Alice'],
    'Age': [25, 30, 35, 40, 28, 32, 45, 29, 31, 25],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney', 'London', 'Paris', 'New York', 'New York'],
    'Salary': [50000, 60000, np.nan, 80000, 55000, 65000, 90000, 52000, 58000, 50000]
}

df = pd.DataFrame(data)

# Use .info() and .describe()
print("--- DataFrame Info ---")
df.info()

print("\n--- DataFrame Describe ---")
print(df.describe())

# Select 'Name' and 'City' columns
print("\n--- Select 'Name' and 'City' ---")
print(df[['Name', 'City']])

# Drop 'City' column
print("\n--- Drop 'City' Column ---")
df_no_city = df.drop(columns=['City'])
print(df_no_city)

# Fill any missing values in 'Salary' column with the mean
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
print("\n--- Fill Missing Salary with Mean ---")
print(df)

# Remove any duplicate rows
df_cleaned = df.drop_duplicates()
print("\n--- After Removing Duplicates ---")
print(df_cleaned)
```
