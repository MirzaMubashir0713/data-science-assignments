```python
''' Problem Statement : Hospital Patient Data Analysis '''
''' 
Context: A hospital maintains patient records including admission details, department, diagnosis, doctor, and bill amount. 
You have two datasets: one with patient info and another with billing details. 
Some patients have blank bill amounts, and there are multiple rows for the same patient due to follow-ups. 

'''

import pandas as pd

# Load datasets
patient_df = pd.read_csv('Patient_Data.csv')
billing_df = pd.read_csv('Billing_Data.csv')


print("------------------------ Task 1 ------------------------")
# 1. Show summary with info()
patient_df.info()


print("\n------------------------ Task 2 ------------------------")
# 2. Select only columns relevant for billing
billing_relevant_df = patient_df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
print(billing_relevant_df)


print("\n------------------------ Task 3 ------------------------")
# 3. Drop administrative columns
patient_df = patient_df.drop(columns=['ReceptionistID', 'CheckInTime'])
print(patient_df.head())


print("\n------------------------ Task 4 ------------------------")
# 4. Groupby to find total bill amount per department
dept_total_bill = patient_df.groupby('Department')['BillAmount'].sum()
print(dept_total_bill)


print("\n------------------------ Task 5 ------------------------")
# 5. Remove duplicate patient records based on PatientID
patient_df = patient_df.drop_duplicates(subset=['PatientID'])
print(patient_df)


print("\n------------------------ Task 6 ------------------------")
# 6. Fill missing BillAmount values with the mean bill amount
mean_bill = patient_df['BillAmount'].mean()
patient_df['BillAmount'] = patient_df['BillAmount'].fillna(mean_bill)
print(patient_df)


print("\n------------------------ Task 7 ------------------------")
# 7. Merge billing dataset with patient dataset on PatientID
merged_df = pd.merge(patient_df, billing_df, on='PatientID', how='inner')
print(merged_df)


print("\n------------------------ Task 8 ------------------------")
# 8. Concatenate additional DataFrame with new patients for current week (row-wise)
new_patients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['Frank', 'Grace'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. Smith', 'Dr. John'],
    'BillAmount': [5500.0, 4800.0]
})

patient_df_updated = pd.concat([patient_df, new_patients], ignore_index=True)
print(patient_df_updated)


print("\n------------------------ Task 9 ------------------------")
# 9. Concatenate new billing category columns (column-wise)
extra_billing_cols = pd.DataFrame({
    'InsuranceCovered': [2000, 1500, 2500, 3000, 1000],
    'FinalAmount': [3000, 3500, 5000, 3200, 4000]
})

final_df = pd.concat([patient_df.reset_index(drop=True), extra_billing_cols], axis=1)
print(final_df)
```
